# A anchor 1

```python
df_NME_2000 = pd.read_csv(r"../Results/Synthetic_Low_Dim_ATT_Experiment/DRPCLNET_SyntheticLowDim_ATT_anchor_1p0_N2000_array_2766660.csv")
df_NME_5000 = pd.read_csv(r"../Results/Synthetic_Low_Dim_ATT_Experiment/DRPCLNET_SyntheticLowDim_ATT_anchor_1p0_N5000_array_2766660.csv")
df_NME_10000 = pd.read_csv(r"../Results/Synthetic_Low_Dim_ATT_Experiment/DRPCLNET_SyntheticLowDim_ATT_anchor_1p0_N10000_array_2766660.csv")
df_NME_15000 = pd.read_csv(r"../Results/Synthetic_Low_Dim_ATT_Experiment/DRPCLNET_SyntheticLowDim_ATT_anchor_1p0_N15000_array_2766660.csv")
df_NME_20000 = pd.read_csv(r"../Results/Synthetic_Low_Dim_ATT_Experiment/DRPCLNET_SyntheticLowDim_ATT_anchor_1p0_N20000_array_2766660.csv")

df_results = pd.concat([df_NME_2000, df_NME_5000, df_NME_10000, df_NME_15000, df_NME_20000], axis = 0)
df_results.tail()
```

```python
plot_title = r"Conditional dose-response (mean with 95% CI), $f_{\mathrm{ATT}}(a,1)$"
high_dim_plot = plot_pcl_convergence_ci(df_results, title = plot_title)
plt.savefig('Figures/DRPCLNET_ATT_Comparison_aprime_1.pdf', format='pdf', bbox_inches = 'tight')
high_dim_plot.show()
```

```python
format_results_for_jupyter_and_latex(df_results, "ATT", precision=6)
```
